```python
import pandas as pd
from fastai.data.block import DataBlock, MultiCategoryBlock
from fastai.data.external import untar_data, URLs
from fastai.vision.augment import RandomResizedCrop
from fastai.vision.data import ImageBlock

path = untar_data(URLs.PASCAL_2012)
```

```python
df = pd.read_csv(path / 'train.csv')
df.head()
```

```python
def get_y_csv(row: pd.Series) -> pd.Series:
    return row['labels']
    
def get_x_csv(row: pd.Series) -> pd.Series:
    return row['fname']
    
get_x_csv(df.iloc[0,:]), get_y_csv(df.iloc[0,:])
```

```python
# dblock = DataBlock()
# 
# dsets = dblock.datasets(df)
# 
# len(dsets.train), len(dsets.valid)
```

```python

def splitter(df: pd.DataFrame):
    train = df.index[~df['is_valid']].to_list()
    valid = df.index[df['is_valid']].to_list()
    return train, valid
splitter(df.head())
```

```python


datablock = DataBlock(
    blocks=[ImageBlock, MultiCategoryBlock],
    get_x=get_x_csv,
    get_y=get_y_csv,
    splitter=splitter,
    item_tfms=RandomResizedCrop(size=128, min_scale=0.35),
)
datablock.datasets(df)
dls = datablock.dataloaders(df)
dls.show_batch()
```
